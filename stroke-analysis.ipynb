```python
age_data = df.groupby("gender")["age"].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6), facecolor="lightyellow")

age_data.plot(
    kind="bar",
    color=["navy", "orange", "green"]
)

plt.title("Average Age by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Age")

plt.show()
```

```python
smoking_data = df.groupby("smoking_status")["age"].count().sort_values(ascending=False)

plt.figure(figsize=(12,6), facecolor="grey")

smoking_data.plot(
    kind="bar",
    color=["navy", "orange", "green", "red"]
)

plt.title("Smoking Status Distribution")
plt.xlabel("Smoking Status")
plt.ylabel("Number of People")

plt.show()
```

```python
top_age = df.sort_values(
    "age",
    ascending=False
).head(10)

plt.figure(figsize=(12,6), facecolor="yellow")

top_age.plot(
    x="id",
    y="age",
    kind="bar",
    color=["navy", "orange", "green", "red", "purple",
           "gold", "brown", "pink", "gray", "olive"]
)

plt.title("Top 10 Oldest People")
plt.xlabel("ID")
plt.ylabel("Age")

plt.show()
```

```python
gender_data = df["gender"].value_counts()

plt.figure(figsize=(8,6), facecolor="lightyellow")

gender_data.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")
plt.ylabel("")

plt.show()
```

```python
age_data = df.groupby("stroke")["age"].mean()

plt.figure(figsize=(12,6), facecolor="lightblue")

age_data.plot(
    kind="line",
    marker="o"
)

plt.title("Average Age by Stroke")
plt.xlabel("Stroke")
plt.ylabel("Average Age")

plt.show()
```

```python
heart_stroke = df.groupby("heart_disease")["stroke"].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6), facecolor="brown")

heart_stroke.plot(
    kind="bar",
    color=["navy", "orange"]
)

plt.title("Stroke Cases by Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Number of Stroke Cases")

plt.show()
```

```python
smoking_data = df["smoking_status"].value_counts()

plt.figure(figsize=(8,6), facecolor="lightblue")

smoking_data.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Smoking Status Distribution")
plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(12,6), facecolor="lightblue")

plt.scatter(
    df["age"],
    df["bmi"]
)

plt.title("Age vs BMI")
plt.xlabel("Age")
plt.ylabel("BMI")

plt.show()
```

```python
plt.figure(figsize=(12,6), facecolor="lavender")

plt.hist(
    df["age"],
    bins=20
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of People")

plt.show()
```

```python
bmi_data = df.groupby("stroke")["bmi"].mean()

plt.figure(figsize=(12,6), facecolor="lightblue")

bmi_data.plot(
    kind="line",
    marker="o"
)

plt.title("Average BMI by Stroke")
plt.xlabel("Stroke")
plt.ylabel("Average BMI")

plt.show()
```

```python
plt.figure(figsize=(12,6), facecolor="lightblue")

plt.scatter(
    df["bmi"],
    df["avg_glucose_level"]
)

plt.title("BMI vs Glucose Level")
plt.xlabel("BMI")
plt.ylabel("Average Glucose Level")

plt.show()
```

```python
bmi_stroke = df.groupby("stroke")["bmi"].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6), facecolor="lavender")

bmi_stroke.plot(
    kind="bar",
    color=["navy", "orange"]
)

plt.title("Average BMI by Stroke")
plt.xlabel("Stroke")
plt.ylabel("Average BMI")

plt.show()
```
